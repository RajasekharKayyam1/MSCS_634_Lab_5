## 1. Data preparation and exploration

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering, DBSCAN
from sklearn.metrics import silhouette_score, homogeneity_score, completeness_score
from scipy.cluster.hierarchy import linkage, dendrogram
from IPython.display import display

sns.set_theme(style="whitegrid")
wine = load_wine()
X = pd.DataFrame(wine.data, columns=wine.feature_names)
y_true = wine.target
print('Dataset shape:', X.shape)
print('True class counts (evaluation only):', np.bincount(y_true))
display(X.head())
X.info()
display(X.describe().round(2))
print('Missing values:', int(X.isna().sum().sum()))
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print('PCA variance explained by first two components:', round(pca.explained_variance_ratio_.sum(), 3))

def plot_clusters(labels, title, ax=None):
    if ax is None:
        _, ax = plt.subplots(figsize=(7, 5))
    labels = np.asarray(labels)
    for label in sorted(np.unique(labels)):
        mask = labels == label
        if label == -1:
            ax.scatter(X_pca[mask, 0], X_pca[mask, 1], c='black', marker='x',
                       s=55, label=f'Noise ({mask.sum()})')
        else:
            ax.scatter(X_pca[mask, 0], X_pca[mask, 1], s=36, alpha=.8,
                       label=f'Cluster {label} ({mask.sum()})')
    ax.set(title=title, xlabel='PCA component 1', ylabel='PCA component 2')
    ax.legend(fontsize=8)
    return ax

plot_clusters(y_true, 'Original Wine Classes (reference only)')
plt.tight_layout(); plt.show()
```

**Interpretation:** Standardization puts the 13 chemical measurements on comparable scales. PCA is used **only for 2D visualization**; clustering and silhouette scores use the full standardized 13-dimensional data.

## 2. Agglomerative hierarchical clustering

```python
hierarchical_rows = []
hierarchical_labels = {}
fig, axes = plt.subplots(1, 3, figsize=(17, 5))
for ax, k in zip(axes, [2, 3, 4]):
    labels = AgglomerativeClustering(n_clusters=k, linkage='ward').fit_predict(X_scaled)
    hierarchical_labels[k] = labels
    hierarchical_rows.append({'algorithm': 'Hierarchical', 'parameters': f'n_clusters={k}',
                              'clusters': len(set(labels)), 'noise_points': 0,
                              'silhouette': silhouette_score(X_scaled, labels),
                              'homogeneity': homogeneity_score(y_true, labels),
                              'completeness': completeness_score(y_true, labels)})
    plot_clusters(labels, f'Ward Hierarchical: k={k}', ax)
plt.tight_layout(); plt.show()
hierarchical_results = pd.DataFrame(hierarchical_rows)
display(hierarchical_results.round(3))
```

```python
Z = linkage(X_scaled, method='ward')
plt.figure(figsize=(15, 6))
dendrogram(Z, truncate_mode='lastp', p=30, show_leaf_counts=True,
           leaf_rotation=45, leaf_font_size=9)
plt.title('Ward-Linkage Dendrogram (last 30 merged groups)')
plt.xlabel('Merged groups / sample indices'); plt.ylabel('Ward linkage distance')
plt.tight_layout(); plt.show()
```

**Dendrogram interpretation:** Each join represents a merge of two groups; taller joins indicate larger linkage distances. A horizontal cut through the dendrogram gives a possible cluster count. Ward linkage tends to create compact groups and requires choosing the number of clusters.

## 3. DBSCAN clustering and parameter sensitivity

```python
dbscan_rows = []
dbscan_labels = {}
settings = [(1.5, 3), (2.0, 5), (2.5, 5), (3.0, 5), (3.5, 5), (3.0, 10)]
fig, axes = plt.subplots(2, 3, figsize=(17, 10))
for ax, (eps, min_samples) in zip(axes.flat, settings):
    labels = DBSCAN(eps=eps, min_samples=min_samples).fit_predict(X_scaled)
    dbscan_labels[(eps, min_samples)] = labels
    n_clusters = len(set(labels) - {-1})
    n_noise = int(np.sum(labels == -1))
    # Silhouette is undefined for a single label and for all-unique labels.
    # Include noise as its own label consistently for this descriptive metric.
    n_labels = len(set(labels))
    sil = (silhouette_score(X_scaled, labels) if 2 <= n_labels < len(labels) else np.nan)
    dbscan_rows.append({'algorithm': 'DBSCAN',
                        'parameters': f'eps={eps}, min_samples={min_samples}',
                        'clusters': n_clusters, 'noise_points': n_noise,
                        'silhouette': sil,
                        'homogeneity': homogeneity_score(y_true, labels),
                        'completeness': completeness_score(y_true, labels)})
    plot_clusters(labels, f'DBSCAN eps={eps}, min_samples={min_samples}', ax)
plt.tight_layout(); plt.show()
dbscan_results = pd.DataFrame(dbscan_rows)
display(dbscan_results.round(3))
```

**Metric caveat:** Homogeneity and completeness above treat DBSCAN noise (`-1`) as a separate predicted label. Silhouette is calculated on all points, including noise as a label, **only when mathematically defined**. This can distort the perceived quality of DBSCAN; interpret scores together with the number of noise points and formed clusters. A setting with zero clusters is not a successful clustering solution even if its external metrics are defined.

## 4. Comparison and insights

```python
results = pd.concat([hierarchical_results, dbscan_results], ignore_index=True)
display(results.round(3))
valid = results[(results['clusters'] >= 2) & results['silhouette'].notna()]
if not valid.empty:
    best = valid.loc[valid['silhouette'].idxmax()]
    print('Highest all-point silhouette among configurations with >=2 clusters:')
    print(best[['algorithm','parameters','clusters','noise_points','silhouette']].to_string())
else:
    print('No configurations with >=2 clusters and defined silhouette.')
print('\nHierarchical (k=3) vs DBSCAN overview:')
display(results[(results['algorithm'] == 'DBSCAN') | (results['parameters'] == 'n_clusters=3')].round(3))
```

### Discussion

- **Hierarchical clustering:** Ward linkage builds a nested grouping and assigns every observation to a cluster. Changing `n_clusters` changes granularity; the dendrogram helps visualize merge distances. A drawback is that a cluster count must be selected, and results can depend on the linkage choice.
- **DBSCAN:** `eps` defines neighborhood radius and `min_samples` defines local density requirements. Smaller `eps` or larger `min_samples` often produces more noise; larger `eps` can merge groups. It can identify outliers and irregularly shaped clusters, but struggles when densities vary or high-dimensional distances become less informative.
- **Evaluation:** Silhouette measures separation and cohesion without class labels; homogeneity measures whether predicted groups contain mainly one true class; completeness measures whether each true class is largely grouped together. These evaluate different qualities and should not be interpreted interchangeably.
- **Dataset-specific conclusion:** Review the **computed table above** to identify which settings create meaningful clusters and how many observations DBSCAN labels as noise. The Wine dataset is relatively small (178 samples) with 13 features, and Ward hierarchical clustering is a reasonable baseline. A high score alone does not prove that a clustering configuration is scientifically useful.
